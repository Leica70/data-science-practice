```python
import seaborn as sns
import pandas as pd

data = {
    'Name': ['Owen', 'Florence', 'Laina', 'Lily', 'William', 'John', 'Mary', 'James', 'Emma', 'Henry'],
    'Sex': ['male', 'female', 'female', 'female', 'male', 'male', 'female', 'male', 'female', 'male'],
    'Age': [22, 38, 26, 35, 35, 45, 29, 52, 31, 41],
    'Pclass': [3, 1, 3, 1, 3, 2, 1, 2, 3, 1],
    'Survived': [0, 1, 1, 1, 0, 0, 1, 0, 1, 1],
    'Fare': [7.25, 71.28, 7.92, 53.10, 8.05, 21.00, 63.50, 18.75, 6.50, 55.90]
}

df = pd.DataFrame(data)
print(df.shape)
df.head()
```

```python
sns.heatmap(df[['Age', 'Pclass', 'Survived', 'Fare']].corr(), 
            annot=True, cmap='coolwarm')
plt.title('Correlation Heatmap')
plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Create full dataset
data = {
    'Name': ['Owen', 'Florence', 'Laina', 'Lily', 'William', 'John', 'Mary', 'James', 'Emma', 'Henry'],
    'Sex': ['male', 'female', 'female', 'female', 'male', 'male', 'female', 'male', 'female', 'male'],
    'Age': [22, 38, 26, 35, 35, 45, 29, 52, 31, 41],
    'Pclass': [3, 1, 3, 1, 3, 2, 1, 2, 3, 1],
    'Survived': [0, 1, 1, 1, 0, 0, 1, 0, 1, 1],
    'Fare': [7.25, 71.28, 7.92, 53.10, 8.05, 21.00, 63.50, 18.75, 6.50, 55.90]
}

df = pd.DataFrame(data)
print("Dataset loaded!")
print(df.shape)
df.head()
```

```python
# Set the style
sns.set_style("whitegrid")
fig, axes = plt.subplots(2, 2, figsize=(12, 10))
fig.suptitle('Titanic Data Analysis', fontsize=16, fontweight='bold')

# Chart 1 - Survivors by Sex
df.groupby('Sex')['Survived'].sum().plot(kind='bar', ax=axes[0,0], color=['pink', 'steelblue'])
axes[0,0].set_title('Survivors by Sex')
axes[0,0].set_xlabel('Sex')
axes[0,0].set_ylabel('Count')

# Chart 2 - Average Fare by Class
df.groupby('Pclass')['Fare'].mean().plot(kind='bar', ax=axes[0,1], color='coral')
axes[0,1].set_title('Average Fare by Class')
axes[0,1].set_xlabel('Class')
axes[0,1].set_ylabel('Average Fare (£)')

# Chart 3 - Age Distribution by Class
sns.boxplot(data=df, x='Pclass', y='Age', ax=axes[1,0], palette='coolwarm')
axes[1,0].set_title('Age Distribution by Class')

# Chart 4 - Survival by Class
sns.countplot(data=df, x='Pclass', hue='Survived', ax=axes[1,1], palette='Set2')
axes[1,1].set_title('Survival by Class')

plt.tight_layout()
plt.show()
```

```python
print("=== TITANIC DATA ANALYSIS SUMMARY ===")
print(f"Total passengers: {len(df)}")
print(f"Total survivors: {df['Survived'].sum()}")
print(f"Survival rate: {df['Survived'].mean() * 100:.1f}%")
print(f"Average age: {df['Age'].mean():.1f}")
print(f"Average fare: £{df['Fare'].mean():.2f}")
print(f"Female survival rate: {df[df['Sex']=='female']['Survived'].mean()*100:.1f}%")
print(f"Male survival rate: {df[df['Sex']=='male']['Survived'].mean()*100:.1f}%")
```
